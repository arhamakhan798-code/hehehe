```python
# ============================================================
# KNN LOAN PREDICTION MODEL
# nice.csv  = TRAINING DATA
# love.csv  = SEPARATE TEST DATA
# 80:20 TRAIN/VALIDATION SPLIT
# ============================================================

import os
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.model_selection import train_test_split, GridSearchCV
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import StandardScaler, OneHotEncoder
from sklearn.neighbors import KNeighborsClassifier
from sklearn.metrics import accuracy_score, confusion_matrix, classification_report

# ============================================================
# 1. LOAD DATA
# ============================================================

train_file = "nice.csv"

# Handles both love.csv and the uploaded "love (1).csv"
if os.path.exists("love.csv"):
    test_file = "love.csv"
elif os.path.exists("love (1).csv"):
    test_file = "love (1).csv"
else:
    raise FileNotFoundError("Please upload love.csv")

train_df = pd.read_csv(train_file)
test_df = pd.read_csv(test_file)

print("================================================")
print("DATASET INFORMATION")
print("================================================")
print("Training file:", train_file)
print("Training shape:", train_df.shape)
print("Testing file:", test_file)
print("Testing shape:", test_df.shape)

# ============================================================
# 2. CHECK MISSING VALUES
# ============================================================

print("\nMissing Values in Training Dataset:")
print(train_df.isnull().sum())

print("\nMissing Values in Testing Dataset:")
print(test_df.isnull().sum())

# ============================================================
# 3. SEPARATE FEATURES AND TARGET
# ============================================================

X = train_df.drop("Loan_Status", axis=1)
y = train_df["Loan_Status"]

X_external = test_df.copy()

# Loan_ID is an identifier, not a useful predictive feature
if "Loan_ID" in X.columns:
    X = X.drop("Loan_ID", axis=1)

if "Loan_ID" in X_external.columns:
    X_external = X_external.drop("Loan_ID", axis=1)

# ============================================================
# 4. 80:20 TRAINING / VALIDATION PARTITION
# ============================================================

X_train, X_val, y_train, y_val = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

print("\n================================================")
print("80:20 PARTITION")
print("================================================")
print("Total records       :", len(X))
print("Training records    :", len(X_train))
print("Validation records  :", len(X_val))
print("Training percentage :", "80%")
print("Validation percentage:", "20%")

# ============================================================
# 5. IDENTIFY NUMERICAL AND CATEGORICAL FEATURES
# ============================================================

numeric_features = X_train.select_dtypes(
    include=["int64", "float64"]
).columns.tolist()

categorical_features = X_train.select_dtypes(
    include=["object"]
).columns.tolist()

print("\nNumerical Features:")
print(numeric_features)

print("\nCategorical Features:")
print(categorical_features)

# ============================================================
# 6. PREPROCESSING
# ============================================================
# Numerical:
#   Missing values -> Median
#   Scaling -> StandardScaler
#
# Categorical:
#   Missing values -> Mode
#   Encoding -> OneHotEncoder
# ============================================================

numeric_pipeline = Pipeline([
    ("imputer", SimpleImputer(strategy="median")),
    ("scaler", StandardScaler())
])

categorical_pipeline = Pipeline([
    ("imputer", SimpleImputer(strategy="most_frequent")),
    ("encoder", OneHotEncoder(handle_unknown="ignore"))
])

preprocessor = ColumnTransformer([
    ("numeric", numeric_pipeline, numeric_features),
    ("categorical", categorical_pipeline, categorical_features)
])

# ============================================================
# 7. KNN MODEL
# ============================================================

knn_pipeline = Pipeline([
    ("preprocessor", preprocessor),
    ("knn", KNeighborsClassifier())
])

# ============================================================
# 8. TEST K = 3, 5, 7
# ============================================================

print("\n================================================")
print("K = 3, 5, 7 COMPARISON")
print("================================================")

k_values = [3, 5, 7]
k_results = {}

for k in k_values:

    model = Pipeline([
        ("preprocessor", preprocessor),
        ("knn", KNeighborsClassifier(
            n_neighbors=k,
            weights="distance"
        ))
    ])

    model.fit(X_train, y_train)

    prediction = model.predict(X_val)

    acc = accuracy_score(y_val, prediction)

    k_results[k] = acc

    print(f"K = {k} --> Accuracy = {acc:.4f} ({acc*100:.2f}%)")

# ============================================================
# 9. HYPERPARAMETER TUNING
# ============================================================

print("\n================================================")
print("HYPERPARAMETER TUNING")
print("================================================")

# Tune:
# n_neighbors
# weights
# distance metric (p)
#
# This searches for a better KNN configuration.

parameter_grid = {
    "knn__n_neighbors": list(range(1, 31)),
    "knn__weights": ["uniform", "distance"],
    "knn__p": [1, 2]
}

grid_search = GridSearchCV(
    estimator=knn_pipeline,
    param_grid=parameter_grid,
    cv=5,
    scoring="accuracy",
    n_jobs=-1
)

grid_search.fit(X_train, y_train)

best_model = grid_search.best_estimator_

print("\nBest Parameters:")
print(grid_search.best_params_)

print("\nBest Cross-Validation Accuracy:")
print(
    f"{grid_search.best_score_:.4f} "
    f"({grid_search.best_score_*100:.2f}%)"
)

# ============================================================
# 10. EVALUATE BEST MODEL ON 20% VALIDATION DATA
# ============================================================

y_val_pred = best_model.predict(X_val)

validation_accuracy = accuracy_score(
    y_val,
    y_val_pred
)

print("\n================================================")
print("BEST KNN MODEL - 20% VALIDATION")
print("================================================")

print(
    f"Validation Accuracy: "
    f"{validation_accuracy:.4f} "
    f"({validation_accuracy*100:.2f}%)"
)

print("\nClassification Report:")
print(classification_report(y_val, y_val_pred))

# ============================================================
# 11. CONFUSION MATRIX
# ============================================================

cm = confusion_matrix(y_val, y_val_pred)

print("\nConfusion Matrix:")
print(cm)

# ============================================================
# 12. CONFUSION MATRIX GRAPH
# ============================================================

plt.figure(figsize=(6, 5))

sns.heatmap(
    cm,
    annot=True,
    fmt="d",
    cmap="Blues",
    xticklabels=["Not Approved (N)", "Approved (Y)"],
    yticklabels=["Not Approved (N)", "Approved (Y)"]
)

plt.title("KNN Confusion Matrix")
plt.xlabel("Predicted Loan Status")
plt.ylabel("Actual Loan Status")
plt.tight_layout()
plt.show()

# ============================================================
# 13. K = 3, 5, 7 ACCURACY GRAPH
# ============================================================

plt.figure(figsize=(7, 5))

plt.plot(
    list(k_results.keys()),
    list(k_results.values()),
    marker="o"
)

plt.xlabel("Number of Neighbors (K)")
plt.ylabel("Accuracy")
plt.title("KNN Accuracy for K = 3, 5, 7")
plt.xticks(k_values)
plt.grid(True)
plt.tight_layout()
plt.show()

# ============================================================
# 14. TRAIN FINAL MODEL USING ALL nice.csv DATA
# ============================================================

print("\n================================================")
print("FINAL MODEL TRAINING")
print("================================================")

final_model = Pipeline([
    ("preprocessor", preprocessor),
    ("knn", KNeighborsClassifier(
        n_neighbors=grid_search.best_params_["knn__n_neighbors"],
        weights=grid_search.best_params_["knn__weights"],
        p=grid_search.best_params_["knn__p"]
    ))
])

# Train on all 614 records
final_model.fit(X, y)

print("Final model trained using all", len(X), "records.")

# ============================================================
# 15. PREDICT love.csv
# ============================================================

external_predictions = final_model.predict(X_external)

# ============================================================
# 16. CREATE PREDICTION FILE
# ============================================================

results = test_df.copy()

results["Predicted_Loan_Status"] = external_predictions

results.to_csv(
    "love_knn_predictions.csv",
    index=False
)

# ============================================================
# 17. DISPLAY TEST PREDICTIONS
# ============================================================

print("\n================================================")
print("PREDICTIONS FOR love.csv")
print("================================================")

print("\nFirst 20 Predictions:")
print(
    results[
        ["Loan_ID", "Predicted_Loan_Status"]
    ].head(20).to_string(index=False)
)

print("\nPrediction Distribution:")
print(
    results["Predicted_Loan_Status"].value_counts()
)

# ============================================================
# 18. FINAL SUMMARY
# ============================================================

print("\n================================================")
print("FINAL SUMMARY")
print("================================================")

print("Training Dataset       : nice.csv")
print("Training Records       :", len(X))
print("Training Split         : 80%")
print("Validation Split       : 20%")
print("External Test Dataset  :", test_file)
print("External Test Records  :", len(X_external))

print("\nK = 3 Accuracy:")
print(f"{k_results[3]*100:.2f}%")

print("K = 5 Accuracy:")
print(f"{k_results[5]*100:.2f}%")

print("K = 7 Accuracy:")
print(f"{k_results[7]*100:.2f}%")

print("\nBest Parameters:")
print(grid_search.best_params_)

print("\nBest Validation Accuracy:")
print(f"{validation_accuracy*100:.2f}%")

print("\nPrediction file saved as:")
print("love_knn_predictions.csv")

print("\nDONE!")
```
